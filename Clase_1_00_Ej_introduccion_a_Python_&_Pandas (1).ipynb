1. Genere una variable con la cantidad de pago recibidos denominada “cant_pagos”.
2. Genere un vector con los valores únicos de los pagos realizados llamado "valores_unicos".
3. Ordene de menor a mayor los pagos en un vector llamado "pagos_ordenados".
4. ¿Cuál es el mínimo valor que se cobró y cual es el máximo valor que se cobró?
5. Genere un nuevo vector denominado "comision" con las comisiones de las cobranzas que se deben abonar al recaudador, recuerde que se cobra el 1.2%.
6. Realice otro vector denominado "comision_pagada" teniendo en cuenta que si la comisión no supera los 8 pesos el valor mínimo de cobranzas es de 8 pesos.
7. Cuántas operaciones se pagaron 8 pesos por no superar ese tope?
8. ¿Cuánto pagué de comisiones y cuanto deberia haber pagado sin no estuviera el tope?

```python
pagos = [100, 300, 400, 500, 6000, 8000, 2012, 1231, 23, 20, 100, 100, 4000, 300, 500, 500, 600, 200, 200, 8000]
```

```python
import numpy as np

pagos = np.array(pagos)
```

```python
# Punto 1
cant_pagos = len(pagos)
print("cant_pagos:", cant_pagos)
```

```python
#Punto 2
valores_unicos = np.unique(pagos)
print("valores_unicos:", valores_unicos)
```

```python
#Punto 3
pagos_ordenados = np.sort(pagos)
print("pagos_ordenados:", pagos_ordenados)
```

```python
#Punto 4
minimo = pagos.min()
maximo = pagos.max()
print("minimo:", minimo, "- maximo:", maximo)
```

```python
#Punto 5
comision = pagos * 0.012
print("comision:", comision)
```

```python
#Punto 6
comision_pagada = np.where(comision < 8, 8, comision)
print("comision_pagada:", comision_pagada)
```

```python
#Punto 7
cant_topeadas = np.sum(comision < 8)
print("Operaciones que pagaron el minimo de 8 pesos:", cant_topeadas)
```

```python
#Punto 8
total_pagado = comision_pagada.sum()
total_sin_tope = comision.sum()
print(f"comisiones pagadas (con tope): {total_pagado:.2f}")
print(f"comisiones sin tope: {total_sin_tope:.2f}")
```

### **Ejercicio 3**
Contexto: Un sistema registró los siguientes datos de clientes: edades, ingresos mensuales en dólares, y géneros.

* ¿Cuál es la edad promedio de los clientes?
* ¿Cuál es el ingreso máximo y el mínimo?
* Calcule el ingreso promedio para mujeres y para varones.
* Genere una lista de booleanos que indique si cada cliente gana más que el promedio general.

```python
import pandas as pd

data = {
    "edad": [25, 34, 22, 45, 52, 23, 31, 40, 29, 37, 28],
    "ingreso": [1800, 2500, 1200, 4000, 5200, 1300, 2100, 3000, 1600, 2700, 2000],
    "genero": ["F", "M", "F", "M", "M", "F", "F", "M", "F", "M", "F"]
}

df = pd.DataFrame(data)

edad_promedio = df["edad"].mean()
print("edad promedio:", edad_promedio)

ingreso_max = df["ingreso"].max()
ingreso_min = df["ingreso"].min()
print("ingreso maximo:", ingreso_max, "- ingreso minimo:", ingreso_min)

ingreso_promedio_mujeres = df.loc[df["genero"] == "F", "ingreso"].mean()
ingreso_promedio_varones = df.loc[df["genero"] == "M", "ingreso"].mean()
print("ingreso promedio mujeres:", ingreso_promedio_mujeres)
print("ingreso promedio varones:", ingreso_promedio_varones)

promedio_general = df["ingreso"].mean()
gana_mas_que_promedio = list(df["ingreso"] > promedio_general)
print("promedio general:", promedio_general)
print("gana mas que el promedio:", gana_mas_que_promedio)
```
